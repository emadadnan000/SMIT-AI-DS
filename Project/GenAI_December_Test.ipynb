```python
from transformers import pipeline
import os
import PyPDF2
```

```python
file_path = "document.txt"

if not os.path.exists(file_path):
    with open(file_path, "w", encoding="utf-8") as f:
        f.write("Artificial Intelligence (AI) is the simulation of human intelligence processes by machines, especially computer systems. "
                "Specific applications of AI include expert systems, natural language processing, speech recognition and machine vision. "
                "The policy of the company states that all employees must complete the AI training module by the end of Q3.")
    print(f"Created sample file: {file_path}")

```

### Task 1: Load the Document

```python
print(f"Reading file: {file_path}")
document_content = ""

if file_path.endswith('.pdf'):
    with open(file_path, 'rb') as pdf_file:
        reader = PyPDF2.PdfReader(pdf_file)
        for page in reader.pages:
            document_content += page.extract_text()
else:
    # Assuming .txt
    with open(file_path, 'r', encoding='utf-8') as txt_file:
        document_content = txt_file.read()

# CLEANUP: Remove newlines to save space
document_content = document_content.replace('\n', ' ').strip()
print(f"Document loaded. Length: {len(document_content)} characters.")
```

```python
qa_pipeline = pipeline("text2text-generation", model="google/flan-t5-base", device=-1)
```

```python
user_question = input("\nEnter your question: ")

prompt = (
    f"Answer the question based ONLY on the context below. "
    f"If the answer is not in the context, say 'Answer not found in the document.'\n\n"
    f"Question: {user_question}\n\n"
    f"Context: {document_content}"
)

# Run the model
# max_length=128: limits the answer size
# truncation=True: automatically cuts the document if it's too big for the model
result = qa_pipeline(prompt, max_length=128, truncation=True)

# Extract the text
final_answer = result[0]['generated_text']
```

```python
print("\n" + "="*30)
print("       FINAL OUTPUT       ")
print("="*30)
print(f"Question:   {user_question}")
print(f"Answer:     {final_answer}")
print(f"Model Used: google/flan-t5-base")
print("="*30)
```
